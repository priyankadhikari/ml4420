# Time Series Analysis

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
from statsmodels.tsa.ar_model import AutoReg
from statsmodels.tsa.statespace.sarimax import SARIMAX
from statsmodels.tsa.arima.model import ARIMA

from sklearn.metrics import mean_absolute_error, mean_squared_error
import datetime as dt
```

```python
avocados = pd.read_csv('spoilage_all.csv')
```

```python
# sum indemnity and acres per month nationally
monthly = (avocados.groupby('month_year')[['Indemnity Amount', 'Determined Acres']].sum().sort_index())

# compute loss rate
monthly['loss_rate'] = monthly['Indemnity Amount'] / monthly['Determined Acres']

monthly = monthly.reset_index()
monthly['month_year'] = pd.to_datetime(monthly['month_year'])
monthly = monthly[monthly['month_year'].dt.year <= 2013]

# Reindex to full monthly calendar (fills gaps with NaN)
monthly = monthly.set_index('month_year').asfreq('MS')

# Interpolate missing months just in case
monthly['loss_rate'] = monthly['loss_rate'].interpolate(method='linear')

print(f'Series length: {len(monthly)} months')
print(f'Date range: {monthly.index.min()} to {monthly.index.max()}')
monthly['loss_rate'].plot(title='Monthly Crop Loss Rate ($/acre)', figsize=(6, 4))
plt.ylabel('Loss Rate ($/acre)')
plt.tight_layout()
plt.show()
```

```python
## check ACF/PACF
 
new = monthly["loss_rate"]

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
plot_acf(new,  lags=30, ax=axes[0],
         title=f"ACF Avocado Loss Rate")
plot_pacf(new, lags=30, ax=axes[1],
          title=f"PACF Avocado Loss Rate")
plt.tight_layout()
plt.show()
```

The oscillating/negative ACF is characteristic of an AR process with a negative coefficient, or mixed ARMA behavior.
The PACF cutting off around lag 1–2 still points to a maybe a low-order AR component.
Neither plot shows a clear cut off or tail off

```python
# chronological test train split
train_size = int(0.8 * len(monthly))
train_data = monthly[:train_size]
test_data  = monthly[train_size:]

y_train = train_data['loss_rate'] 
y_test  = test_data['loss_rate']
```

```python
# Visualize the training data
plt.plot(train_data['loss_rate'], label='(Training) Loss Rate')
plt.xlabel('Month')
plt.ylabel('Loss Rate ($/acre)')
plt.legend()
plt.title('Avocado Loss Rate over Time')
plt.show();
```

```python
arima_model = ARIMA(y_train, order=(20, 1, 20), trend='n')
arima_results = arima_model.fit()

arima_results.summary()
```

```python
abs(arima_results.arroots)
```

```python
arima_pred = arima_results.predict(start=test_data.index[0], end=test_data.index[-1])
pred_data = pd.DataFrame({'Month': test_data.index, 'pred loss_rate': arima_pred})
pred_data.set_index('Month', inplace=True)
pred_data.head()
```

```python
# Make the plot
plt.plot(test_data['loss_rate'], label='Actual Loss Rate')
plt.plot(pred_data['pred loss_rate'], label='Predicted loss rate', linestyle='--')
plt.xlabel('Month')
plt.ylabel('Loss Rate ($/acre)')
plt.legend()
plt.title('ARIMA(20, 1, 20) Model')
plt.xticks(rotation = 45)
plt.show();
```

```python
arima_results.plot_diagnostics(figsize=(9,10));
```

```python
mae = mean_absolute_error(y_test, arima_pred)
rmse = np.sqrt(mean_squared_error(y_test, arima_pred))
print(f'Mean Absolute Error: {mae:.2f}')
print(f'Root Mean Squared Error: {rmse:.2f}')
```
